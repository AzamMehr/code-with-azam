# Welcome to Jupyter, Java Developer!

If you are used to IntelliJ or Eclipse, Jupyter Notebooks will feel unusual at first — but they are one of Python's killer features for learning, data exploration, and rapid prototyping.

## Key differences from Java development

| Java workflow | Jupyter workflow |
|---|---|
| Write `.java` → `javac` → `java` | Write in a cell → `Shift+Enter` |
| One compilation unit | Each cell runs independently |
| Variables scoped to method/class | Variables persist across cells (shared kernel state) |
| Errors at compile time | Errors shown inline, immediately |
| Full restart to reset state | **Kernel → Restart** to clear all variables |

## The mental model

Think of a notebook as a **live REPL session that you can save**.  
Cells can be re-run in any order, but the recommended practice is top-to-bottom.  
The number in `In [3]:` tells you the execution order — not the cell position.

> **Tip:** Use **Kernel → Restart & Run All** before sharing a notebook to verify it runs cleanly from top to bottom.

```python
print("Hello from Jupyter! If you see this, your setup works.")
```

```python
# Variables defined here persist for the rest of the notebook session.
# In Java you'd write:  String language = "Python";  int experienceYears = 10;
# Python uses snake_case and needs no type declaration (though we can add hints).

language: str = "Python"
experience_years: int = 10
```

```python
# f-strings are Python's equivalent of String.format() or text blocks — but nicer.
# Java: System.out.printf("Learning %s with %d years of experience%n", language, experienceYears);
# Python:
print(f"Learning {language} with {experience_years} years of experience.")
```

```python
# Jupyter auto-displays the value of the LAST expression in a cell (no print needed).
# This is unique to notebooks — plain .py scripts do NOT do this.
# Java equivalent: (int) Math.pow(2, 10)

2 ** 10  # In Java: (int) Math.pow(2, 10)
```

```python
import sys
from datetime import datetime

# sys.version gives the full version string, including build info.
print(f"Python version : {sys.version}")
print(f"Current time   : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
# List comprehensions — Python's answer to Java streams for simple transforms.
#
# Java (streams):
#   List<Integer> squares = IntStream.rangeClosed(1, 5)
#                                    .map(i -> i * i)
#                                    .boxed()
#                                    .collect(Collectors.toList());
#
# Python (list comprehension):
squares = [i ** 2 for i in range(1, 6)]

print(f"Squares 1-5: {squares}")
```
