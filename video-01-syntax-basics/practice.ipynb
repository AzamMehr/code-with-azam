### Exercise 3.2 -- String Slicing (No Java Equivalent!)

**Java:**
```java
String s = "Hello, World!";
String sub = s.substring(0, 5);  // "Hello"
// Reversing requires: new StringBuilder(s).reverse().toString()
```

**Task:** Use Python slicing `s[start:stop:step]` to get the first 5 chars, last 6 chars, and reverse the whole string. Each answer is ONE expression.

````python
s = "Hello, World!"

first_five = "REPLACE"  # hint: s[?:?]
last_six   = "REPLACE"  # hint: s[-?:]
reversed_s = "REPLACE"  # hint: s[::-?]

print(f"First 5: {first_five}")
print(f"Last 6:  {last_six}")
print(f"Reversed: {reversed_s}")

# Expected output:
# First 5: Hello
# Last 6:  World!
# Reversed: !dlroW ,olleH
````

<details>
<summary>Solution</summary>

```python
first_five = s[:5]
last_six   = s[-6:]
reversed_s = s[::-1]
```

**Why it works:** Slicing syntax is `s[start:stop:step]`. Negative indices count from the end (`-6` = 6th from last). `[::-1]` means "whole string, step backwards by 1" which reverses it. This is one of Python's most loved features and has no Java equivalent.
</details>


**Task:** Translate each line to Python. Hint: Python's `join()` is called on the separator, not the list.

````python
items = "apple,banana,cherry"

parts   = "REPLACE"  # hint: items.split(...)
joined  = "REPLACE"  # hint: " | ".join(...)  <-- separator goes FIRST
divider = "REPLACE"  # hint: "=" * 30

print(f"Parts: {parts}")
print(f"Joined: {joined}")
print(divider)

# Expected output:
# Parts: ['apple', 'banana', 'cherry']
# Joined: apple | banana | cherry
# ==============================
````

### Exercise 4.3 -- Truthiness Discovery

**Java:**
```java
// In Java, only boolean works in if conditions
// if (0) { }   // Compile error!
// if ("") { }  // Compile error!
```

**Task:** Just run the cell below. Predict the results first, then verify. No code to write -- this is a discovery exercise.

````python
# Which of these are truthy? Which are falsy?
# Predict first, then run to verify!
test_values = [0, 1, "", "hello", [], [1, 2], None, False, True, {}, 0.0]

for val in test_values:
    result = "truthy" if val else "falsy"
    print(f"{str(val):>10} -> {result}")

print("\nIn Java, NONE of these work in an if condition except True/False.")
print("In Python, all of them work. Empty/zero versions are falsy.")
````

<details>
<summary>Discovery Results</summary>

**Falsy:** `0`, `""` (empty string), `[]` (empty list), `{}` (empty dict), `None`, `False`, `0.0`

**Truthy:** `1`, `"hello"`, `[1, 2]`, `True`, and any non-zero / non-empty value

**Rule:** Empty collections, zero numbers, None, and False are falsy. Everything else is truthy.

In practice this lets you write `if items:` instead of `if len(items) > 0:` -- a very common Pythonic pattern.
</details>

**Task:** Fix the `enumerate` call to iterate over the actual `fruits` list (not the placeholder).

````python
fruits = ["apple", "banana", "cherry"]

for i, fruit in enumerate(["REPLACE"]):  # REPLACE ["REPLACE"] with fruits
    print(f"{i}: {fruit}")

# Expected output:
# 0: apple
# 1: banana
# 2: cherry
````

### Exercise 5.3 -- zip() Iterates Two Lists Together

**Java:**
```java
String[] names = {"Alice", "Bob", "Charlie"};
int[] ages = {30, 25, 35};
for (int i = 0; i < names.length; i++) {
    System.out.println(names[i] + " is " + ages[i]);
}
```

**Task:** Fix the `zip` call to pair up the real `names` and `ages` lists.

````python
names = ["Alice", "Bob", "Charlie"]
ages  = [30, 25, 35]

for name, age in zip(["REPLACE"], [0]):  # REPLACE with zip(names, ages)
    print(f"{name} is {age}")

# Expected output:
# Alice is 30
# Bob is 25
# Charlie is 35
````

<details>
<summary>Solution</summary>

```python
for name, age in zip(names, ages):
    print(f"{name} is {age}")
```

**Why it works:** `zip()` pairs elements from multiple iterables together. No index needed. If the lists are different lengths, `zip()` stops at the shortest one.
</details>

## Section 6: Operators

### Exercise 6.1 -- Operators That Look Different

**Java:**
```java
int result1 = 7 / 2;               // 3  (integer division)
double result2 = Math.pow(2, 10);  // 1024.0
boolean check = !true && false;    // false
```

**Task:** Replace each placeholder with the correct Python operator expression.

````python
result1 = "REPLACE"  # integer division of 7 by 2  (hint: use //)
result2 = "REPLACE"  # 2 to the power of 10        (hint: use **)
check   = "REPLACE"  # not True and False           (hint: use words, not symbols)

print(f"7 integer-div 2   = {result1}")
print(f"2 to the power 10 = {result2}")
print(f"not True and False = {check}")

# Expected output:
# 7 integer-div 2   = 3
# 2 to the power 10 = 1024
# not True and False = False
````

<details>
<summary>Solution</summary>

```python
result1 = 7 // 2
result2 = 2 ** 10
check   = not True and False
```

**Why it works:** `//` is integer (floor) division -- NOT a comment! `**` is exponentiation. `and`, `or`, `not` replace `&&`, `||`, `!`. Note: `2 ** 10` returns `int` not `float`, unlike Java's `Math.pow()`.

**Common Java-brain mistake:** Writing `7 / 2` in Python gives `3.5`, not `3`. Always use `//` when you want integer division.
</details>

### Exercise 6.2 -- The `in` Operator (No Java Equivalent)

**Java:**
```java
List<String> roles = List.of("admin", "editor", "viewer");
if (roles.contains("admin")) { System.out.println("Has admin access"); }
String email = "user@example.com";
if (email.contains("@")) { System.out.println("Valid email format"); }
```

**Task:** Replace the placeholder strings with the correct values. The `in` operator works on both lists and strings.

````python
roles = ["admin", "editor", "viewer"]
email = "user@example.com"

if "REPLACE" in roles:  # REPLACE with "admin"
    print("Has admin access")

if "REPLACE" in email:  # REPLACE with "@"
    print("Valid email format")

# Expected output:
# Has admin access
# Valid email format
````

<details>
<summary>Solution</summary>

```python
if "admin" in roles:
    print("Has admin access")

if "@" in email:
    print("Valid email format")
```

**Why it works:** The `in` operator works on lists, strings, dicts (checks keys), sets, and any iterable. Java requires `.contains()` on the object. Python puts the needle first: `needle in haystack`. You can also use `not in`: `if "banned" not in roles:`.
</details>
